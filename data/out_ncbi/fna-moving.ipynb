```python
import os
import shutil
import pandas as pd
```

```python
# === Variables you will fill later ===
main_folder = 'ncbi_dataset/data'   # where all the genome folders are
destination_folder = '../out_fna'   # where you want to save the fna files
tsv_path = 'ncbi_dataset/data/data_summary.tsv'  # path to the tsv file
```

```python
# === Load the TSV without headers ===
df = pd.read_csv(tsv_path, sep='\t', header=None)

df[0] = df[0].str.replace('.', '_')

# Create a mapping: folder name (column 5) -> new name (column 0 with spaces replaced)
rename_dict = dict(zip(df[6], df[0].str.replace(' ', '_')))
```

```python
# === Ensure destination folder exists ===
os.makedirs(destination_folder, exist_ok=True)

# === Process each subfolder inside main_folder ===
for entry in os.scandir(main_folder):
    if entry.is_dir():
        folder_name = entry.name
        new_name_base = rename_dict.get(folder_name)

        if new_name_base is None:
            print(f"No rename entry found for folder '{folder_name}'. Skipping.")
            continue

        # Look for .fna files inside this folder
        for file in os.listdir(entry.path):
            if file.endswith('.fna'):
                old_path = os.path.join(entry.path, file)
                new_filename = f"{new_name_base}.fna"
                new_path = os.path.join(destination_folder, new_filename)
                
                shutil.copy2(old_path, new_path)  # Or use shutil.move to move instead of copy
                print(f"Moved and renamed: {old_path} -> {new_path}")
```
